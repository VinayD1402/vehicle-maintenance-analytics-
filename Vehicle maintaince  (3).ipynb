## 7.2 Data Timing and Leakage Consideration

Before building the predictive model, it is important to consider when each feature becomes available relative to the maintenance decision.

### Scenario A — Information available before the maintenance decision

Vehicle arrives / is assessed
        ↓
Reported Issues
Brake Condition
Tire Condition
Battery Status
        ↓
ML model makes prediction
        ↓
"Maintenance required?"
        ↓
Need_Maintenance = 1

In this scenario, these variables are legitimate input features because they are available at the time of maintenance-risk assessment.

### Scenario B — Information recorded after the maintenance decision

### Sensitivity Test Using Reduced Features

Three highly predictive features—`Reported_Issues`, `Brake_Condition`, and `Battery_Status`—are temporarily excluded to evaluate their influence on model performance.

The same stratified train-test split and evaluation metrics are used to compare Decision Tree and Random Forest models. A substantial drop in performance would indicate that these features contribute heavily to the original results.

This is a diagnostic experiment, not a final decision to remove the features. Their suitability must also be judged by whether they are genuinely available before the maintenance prediction is made.

````python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 1. Temporarily exclude suspiciously predictive features
test_features = [
    col for col in features
    if col not in [
        "Reported_Issues",
        "Brake_Condition",
        "Battery_Status"
    ]
]

X_test_features = df[test_features]
y = df["Need_Maintenance"]

# 2. Identify numerical and categorical columns
test_numeric = [
    col for col in test_features
    if col in numeric_features
]

test_categorical = [
    col for col in test_features
    if col in categorical_features
]

# 3. Split the data into training and testing sets
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_test_features,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 4. Prepare numerical and categorical data
test_preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), test_numeric),
        ("cat", OneHotEncoder(handle_unknown="ignore"), test_categorical)
    ]
)

# 5. Define the models
models = {
    "Decision Tree (Reduced Features)": DecisionTreeClassifier(
        random_state=42
    ),
    "Random Forest (Reduced Features)": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
}

# 6. Train and evaluate both models
results = []

for name, model in models.items():
    pipeline = Pipeline([
        ("preprocessor", test_preprocessor),
        ("model", model)
    ])

    pipeline.fit(X_train_s, y_train_s)
    predictions = pipeline.predict(X_test_s)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test_s, predictions),
        "Precision": precision_score(
            y_test_s, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test_s, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_test_s, predictions, zero_division=0
        )
    })

# 7. Display the comparison
reduced_model_comparison = pd.DataFrame(results)

print(reduced_model_comparison.round(4))
````

### Reduced-Feature Model Comparison

After excluding `Reported_Issues`, `Brake_Condition`, and `Battery_Status`, the Random Forest achieved 80.93% accuracy, 81.01% precision, 99.88% recall, and an 89.46% F1-score. The Decision Tree achieved 69.54% accuracy and 79.62% recall.

The Random Forest performed better overall and identified almost all vehicles requiring maintenance. However, the original models' perfect scores require further investigation. These results are preliminary and do not establish whether the excluded features cause data leakage.

### Five-Fold Cross-Validation

Five-fold stratified cross-validation divides the dataset into five parts while preserving the approximate maintenance/non-maintenance ratio in each fold. The models are trained and evaluated five times, using a different fold for validation each time.

The mean metrics provide a more stable estimate of performance than relying on a single train-test split. However, cross-validation cannot fix data leakage if the input features themselves contain information unavailable at prediction time.

The reduced feature set is used for this experiment to evaluate model stability after excluding three highly predictive features.

````python
# check cross-validation performance

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Use the reduced feature set from the previous test
X_cv = df[test_features]
y_cv = df["Need_Maintenance"]

# Preprocessing
preprocessor_cv = ColumnTransformer([
    ("num", StandardScaler(), test_numeric),
    ("cat", OneHotEncoder(handle_unknown="ignore"), test_categorical)
])

# Models
models_cv = {
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
}

# Five-fold stratified cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    "accuracy": "accuracy",
    "precision": "precision",
    "recall": "recall",
    "f1": "f1"
}

cv_results = []

for name, model in models_cv.items():
    pipeline = Pipeline([
        ("preprocessor", preprocessor_cv),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X_cv,
        y_cv,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": name,
        "Accuracy_Mean": scores["test_accuracy"].mean(),
        "Precision_Mean": scores["test_precision"].mean(),
        "Recall_Mean": scores["test_recall"].mean(),
        "F1_Mean": scores["test_f1"].mean()
    })

cv_comparison = pd.DataFrame(cv_results)
print(cv_comparison.round(4))
````

### Five-Fold Cross-Validation Results

Five-fold stratified cross-validation was used to evaluate the reduced-feature Decision Tree and Random Forest models.

The Random Forest achieved mean accuracy of 80.96%, precision of 81.01%, recall of 99.93%, and F1-score of 89.48%. The Decision Tree achieved mean accuracy of 69.98% and F1-score of 81.20%.

The Random Forest performed better overall, particularly in recall. However, cross-validation alone does not rule out data leakage. The model's features and their availability at prediction time must be verified before selecting it as the final model.

### Inspecting Prediction Probabilities and Classification Thresholds

The Random Forest produces a probability estimate for each vehicle. By default, a probability of 0.50 or higher is classified as maintenance required.

Inspecting the probability distribution and applying alternative thresholds helps us understand whether the model can distinguish vehicles that require maintenance from those that do not.

A higher threshold generally reduces positive predictions but may also increase missed maintenance cases. The final threshold should reflect the cost of unnecessary maintenance versus the cost of missing a genuine maintenance requirement.

````python
## : inspect prediction probabilities
import pandas as pd
from sklearn.metrics import precision_recall_curve

# Get predicted probabilities for the Maintenance class
maintenance_probabilities = final_rf_pipeline.predict_proba(X_test_s)[:, 1]

# Inspect the probability distribution
print("Predicted probability summary:")
print(pd.Series(maintenance_probabilities).describe().round(4))

# See how many predictions fall below common thresholds
for threshold in [0.50, 0.60, 0.70, 0.80, 0.90]:
    predictions_at_threshold = (
        maintenance_probabilities >= threshold
    ).astype(int)

    print(
        f"Threshold: {threshold:.2f} | "
        f"Predicted maintenance: {predictions_at_threshold.sum()} | "
        f"Predicted no maintenance: {(predictions_at_threshold == 0).sum()}"
    )
````

### Classification Threshold Evaluation

The training data is divided into a smaller training set and a validation set. The Random Forest is fitted on the smaller training set, and different probability thresholds are evaluated on the validation set.

The comparison includes maintenance precision and recall, recall for vehicles that do not require maintenance, F1-score, false positives, and false negatives.

The validation set helps us select a threshold without repeatedly tuning against the final test set. The best threshold depends on the business costs of unnecessary maintenance and missed maintenance requirements.

````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score, accuracy_score

# Split the existing training data into training and validation sets
X_train_v, X_val, y_train_v, y_val = train_test_split(
    X_train_s,
    y_train_s,
    test_size=0.20,
    random_state=42,
    stratify=y_train_s
)

# Build and train the reduced-feature Random Forest
validation_rf = Pipeline([
    ("preprocessor", test_preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ))
])

validation_rf.fit(X_train_v, y_train_v)

# Probability of requiring maintenance
val_probabilities = validation_rf.predict_proba(X_val)[:, 1]

# Evaluate different thresholds on validation data
threshold_results = []

for threshold in [0.50, 0.60, 0.70, 0.80, 0.90]:
    val_predictions = (val_probabilities >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        y_val, val_predictions, labels=[0, 1]
    ).ravel()

    threshold_results.append({
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_val, val_predictions),
        "Maintenance_Precision": precision_score(
            y_val, val_predictions, zero_division=0
        ),
        "Maintenance_Recall": recall_score(
            y_val, val_predictions, zero_division=0
        ),
        "No_Maintenance_Recall": tn / (tn + fp) if (tn + fp) else 0,
        "F1_Score": f1_score(
            y_val, val_predictions, zero_division=0
        ),
        "False_Positives": fp,
        "False_Negatives": fn
    })

threshold_comparison = pd.DataFrame(threshold_results)
print(threshold_comparison.round(4).to_string(index=False))
````

### Classification Threshold Comparison

The validation results show that increasing the classification threshold reduces false-positive maintenance predictions but increases false negatives. At a threshold of 0.50, maintenance recall is 99.92%, but no-maintenance recall is 0%. At a threshold of 0.70, maintenance recall decreases to 84.52%, while no-maintenance recall improves to 24.61%.

A threshold of 0.70 is retained as a provisional candidate for further evaluation, not a final operational setting. The final threshold should be chosen according to the relative costs of unnecessary maintenance and missed maintenance requirements. Safety-critical maintenance decisions should not rely solely on model predictions.
